### SET'S METHOD 

#### 1 . ADD ( ) - we can add element ussing this method

```python
a={10,20,30.5,60}
a.add(500)
print(a)
```

```python
a1={1,2,3,5,6}
a1.add("vishu")
print(a1)
```

#### 2. UPDATE ( ) - using this method we cann add elements of one set onto the oher set

```python
b={10,20,30,40}
b1={1,2,3}
b.update(b1)
print(b)
```

```python
b22={"vishu",5,9}
b23={96,63}
b23.update(b22)
print(b23)
```

#### 3. REMOVE ( ) - we can remove particular element from the set

```python
a={10,20,30.5,60}
a.remove(10)
print(a)
```

```python
a={10,20,30.5,60}
a.remove(10000)
print(a)
```

```python
a1={1,2,3,5,6}
a1.add("vishu")
print(a1)
a1.remove("vishu")
print(a1)
```

#### 4. DISCARD ( ) - Remove the particular item
#### in discard method if particular element is not in set it,s return nothing , and in remove method its, gives an error

```python
a={10,20,30.5,60}
a.discard(1000)
print(a)
```

```python
a1={1,2,3,5,6}
a1.add("vishu")
print(a1)
a1.discard("vishu")
print(a1)
```

#### 5. CLEAR - it removes the all element from the set and make set empty 

```python
a={10,20,30.5,60}
a.clear()
print(a)
```

#### 6. COPY ( )  -  its make the copy of set

```python
a={10,20,30.5,60}
a1=a.copy()
print(a)
print(a1)
```
