### SETS { }

##### Set's is unordered , sequence not maintain here

#### Sets is defined using { } braces and element seprated by comma's and encloed in  { } braces

#### 1. HOMOGENEOUS ELEMENT { }

##### Here , we add only 1 data Types Element 

```python
a = { 10,20,30,40,50,60}
print(a)
print(type(a))
print(id(a))
```

### 2. HETROGENEOUS ELEMENT { }

##### here, we add multiple data Types 

```python
a={1,1.2,2+5j,"vishu",True,False}
print(a)
print(type(a))
```

```python
b={25,25.6,96,2+6j,"heyy"}
print(b)
```

### LOOPING IN SET'S

#### 1. FOR LOOP

```python
s2={20,10,40,50,60}
for i in s2:
    print(i)
```

```python
a1={1,2,3,5,4,1,2,3,4,5,12.5,56,89,"vishu","vishu"}
for i in a1:
    print(i)
```

#### 2. ENUMERATE ( )

```python
s3={10,50,60,70,80,90,50,60,1,2,10}
for index,value in enumerate(s3):
    print(index,value)
```

```python
a1={1,2,3,5,4,1,2,3,4,5,12.5,56,89,"vishu","vishu"}
for index,value in enumerate(a1):
    print(index,value)
```
