### 3. ALIASING ( )

#### Both variable Refer To 1 Memory Address

```python
a={10,20}
b=a
c=b
d=a
b={20,50,60,80}
print(d)
print(id(a))
print(id(b))
print(id(c))
print(id(d))
print(b)
```

### 4. Clonning ( )

```python
v={10,1,2,3,4,5}
f=v.copy()
print(f)
print(v)
print(id(f))
print(id(v))
```

#### 6. identity - is operator

```python
a={10,20}
b=a
c=b
d=a
n={10}
print(a is b)
print(b is c)
print(d is a)
print(c is a)
print(n is a)
```
