### PYTHON BASIC DATA-TYPES


#### 1.INTEGER (INT)

```python
decimal_num=10
print(decimal_num)
```

#### binary representation

```python
binary_num=0b1010
print(binary_num)
```

#### 3.OCTAL REPRESENTATION

```python
octal_num=0o12
print(octal_num)
```

#### 4.HEXA_DECIMAL_NUM

## FLOAT

### 1.DECIMAL NOTATION

```python
a=102.2
print(a)
print(type(a))
```

### 2.EXPONENTIAL NOTATION

```python
var1=6.02e23
print(var1)
print(type(var1))
```

### COMPLEX -  IMAGINARY

```python
a=5+10j
print(a)
print(type(a))
print(a.real)
print(a.imag)   
```

### BOOLEAN [bool]

```python
marks=78
has_passed=marks>35
print(has_passed)
print(type(has_passed))
marks=27
has_passed=marks>35
print(has_passed)
```

### STRING (STR)

```python
city='pune'
course="python"
course_desc=''' this is an
amazing course to learn python programming language'''

print(city)
print(course)
print(course_desc)
print(type(city))
```

### Variable Assaignment (Dynamic Typing)

```python
age=25
print(age)
print(type(age))

a1="twenty five"
print(a1)
print(type(a1))

age1=25.6
print(age1)
print(type(age1))

age56=True
print(age56)
print(type(age56))

age2=1+2j
print(age2)
print(type(age2))
```

### Multiple Variable Assignment

```python
length=12
breadth=8
height=5

length,breadth,height=12,8,5
print(length)
print(breadth)
print(height)

print(length,breadth,height)

length=breadth=height=8
print(length,breadth,height)
```
