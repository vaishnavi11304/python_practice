## String Slicing

```python
str1="jai ganesh"
print(str1[0:3:1])
print(str1[0:10:2])
print(str1[0:-8:1])
print(str1[::])
```

## Reverse The String

```python
str2="jai ganesh"
print(str2[::-1])
```
