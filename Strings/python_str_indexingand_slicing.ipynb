### Strings

#### SINGLE QUOTES

```python
greetings='jay ganesh'
print(greetings)
print(type(greetings))

```

#### DOUBLE QUOTES

```python
greetings1="jay ganesh"
print(greetings1)
print(type(greetings1))
```

#### TRIPLE QUOTES - we use it for multiline comments

```python
greetings3='''jay ganesh'''
print(greetings3)
print(type(greetings3))
```

### MULTILINE STRINGS

```python
greetings4='''hello , my name is vaishnavi
i am new in python learning course\
happy to learn
python..... 
it's fun to learn '''
print(greetings4)
```

## INDEXING 

### Poisitive indexing

```python
str1="jai ganesh"
print(str1)
print(str1[4])
```

### Negative Indexing

```python
str2="jai ganesh"
print(str2)
print(str2[-4])
```
