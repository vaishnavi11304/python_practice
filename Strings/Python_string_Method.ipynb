### 1.Isalpha ()

```python
a1="helloworld"
print(a1.isalpha())
```

### 2.Is digit()

```python
a2="1231"
print(a2.isdigit())
```

### 3.isalnum ()

```python
a3="hello1234"
print(a3.isalnum())
```

### 4.Is lower ()

```python
str10="hello world"
print(str10.islower())
```

### 5.is upper ()

```python
str11="HELLO WORLD"
print(str11.isupper())
```

### 6.startswith ()

```python
str12="hello world 11"
print(str12.startswith("hello"))
```

### 7.Endswith ()

```python
str13="hello world 11"
print(str13.endswith("11"))
```
