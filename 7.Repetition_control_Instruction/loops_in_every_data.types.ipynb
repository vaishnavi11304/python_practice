#### while loop - ( )

### while → repeats based on a condition

### 1. LIST ( )

```python
numbers = [10, 20, 30]
i = 0

while i < len(numbers):
    print(numbers[i])
    i += 1
```

### 2. TUPLE

```python
numbers = (10, 20, 30)
i = 0

while i < len(numbers):
    print(numbers[i])
    i += 1
```

### 3. STRING ( )

```python
name = "Vishu"
i = 0

while i < len(name):
    print(name[i])
    i += 1
```

#### WE don't use while loop in set because its unorderd

#### IF YOU WANT TO USE WHILE LOOP IN SET U NEED TO 1ST CONVERT AND THAN YOU CAN USE IT

```python
numbers = {10, 20, 30,60,90,100}
numbers = list(numbers)

i = 0

while i < len(numbers):
    print(numbers[i])
    i += 1
```

## 2. FOR LOOP ( )

### for → naturally goes through items

### 1. LIST ( )

```python
numbers = [10, 20, 30,20,50,90,80,700]

for x in numbers:
    print(x)
```

## 2. TUPLE ( )

```python
numbers = (10, 20, 30,1,2,3,5)
for x in numbers:
    print(x)

```

## 3. STRING ( )

```python
name = "Vishu"
for x in name:
    print(x)
```

## 4. SET ( )

```python
numbers = {10, 20, 30}
for x in numbers:
    print(x)
```
