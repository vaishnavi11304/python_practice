### Repetation Control Statement

### 1. While Loop - 

#### WHILE = CONDITION IS TRUE
##### WHILE = CONDITION

#### Keep going while...

#### Do I need to keep doing something
WHILE a condition is true  


       ↓
      YES
       ↓
   WHILE loop


```python
count=0
while (count<=5):
    print(count)
    count=count+1
```

```python
i = 0

while i < 4:
    print(i)
    i += 1

# Keep going WHILE i < 4
```

```python
basket = ["apple", "banana", "mango"]

while basket:
    fruit = basket.pop()
    print(fruit)

# Keep taking things WHILE the basket isn't empty
```

### 2.For Loop ( )

```python
list1=[1,"ganesh",4,4.23,4+1j,True,000,12365]
print(list1)
for i in list1:
    print(i)
```

```python
str1="ganesh" 
for i in str1:
    print(i)
```

### Range ( )  - Built In Function

```python
for g in range(1,11,3):
    print(g)
```

### Break Statement ( )

```python
for i in range(1,50,1):
    print(i)
    if i==6:
        break
    print(i)
```

### Continue ( )

```python
for i in range(1,20,1):
    #print(i)
    if i==6:
        continue
    print(i)
```
