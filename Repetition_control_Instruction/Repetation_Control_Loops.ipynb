### Repetation Control Statement

### 1. While Loop - 

#### WHILE = CONDITION IS TRUE
##### WHILE = CONDITION

#### Keep going while...

#### Do I need to keep doing something
WHILE a condition is true  


       ↓
      YES
       ↓
   WHILE loop


```python
count=0
while (count<=5):
    print(count)
    count=count+1
```

```python
i = 0

while i < 4:
    print(i)
    i += 1

# Keep going WHILE i < 4
```

```python
basket = ["apple", "banana", "mango"]

while basket:
    fruit = basket.pop()
    print(fruit)

# Keep taking things WHILE the basket isn't empty
```

```python
i = 0

while i < 6:
    i += 1

    if i == 3:
        continue

    print(i)
```

### 2.For Loop ( )

#### FOR loop = Give me each item

```python
list1=[1,"ganesh",4,4.23,4+1j,True,000,12365]
print(list1)
for i in list1:
    print(i)

# Python, give me each item from list
```

```python
str1="ganesh" 
for i in str1:
    print(i)
```

### Range ( )  - Built In Function

#### for i, x in enumerate(data) → index + item

```python
for g in range(1,11,3):
    print(g)

# Start at 1 → keep jumping by 3 → stop before reaching 11.
```

### Break Statement ( )

#### you will type break wherever it can stop from those element not going on next 

```python
for i in range(1,50,1):
    print(i)
    if i==6:
        break
    print(i)
```

### Continue ( )

#### in continue statement after you type continue it will skip from that element and continue from next

### continue = skip this iteration

```python
for i in range(1,20,1):
    #print(i)
    if i==6:
        continue
    print(i)
```
