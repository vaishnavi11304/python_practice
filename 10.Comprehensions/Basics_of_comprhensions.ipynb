### Comprhension

#### Comprehension: A short way to write a loop + operation/code in one line = comprehension

#### List Comprehension . [ ]

#### Without Comprhension [ ]

```python
''' usually we writhe the code without comprhension in multiline but using comprhension we write
a code in one line and a shortly'''


numbers = [1, 2, 3, 4, 5]
result = []

for x in numbers:
    result.append(x * 2)

print(result)
```

#### With Comprhension

```python
numbers = [1, 2, 3, 4, 5]
result = [a * 2 for a in numbers]

print(result)

# Here:

# a → represents the current number
# * → means multiplication
# 2 → the number you're multiplying by
```
