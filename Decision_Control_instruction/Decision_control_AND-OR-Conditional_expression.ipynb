##  AND , OR , CONDITIONAL EXPRESSION

#### 1) Logical Operator 
#### 1. AND ( )

```python
age =2
nationality="indian"

if (age>=18) and (nationality=="indian"):
    print("You are eligible to vote")
else:
    print("you are not eligible for voting")
```

#### 2. OR ( )

```python
weekend=True
holiday=False
if weekend or holiday:
    print("You can sleep in")
else:
    print("Get up early")
```

#### Conditional Expression ( )

### With the multiple line 

```python
age=24
if age>=18:
    print("You are eligible to vote")
else:
    print("you are not eligible for voting")

```

### In a Single Line

```python
age=24
"you are eligible to vote" if age>=18 else "you are not eligible for voting"
```

#### 1.ALL ( ) - Built in Function

```python
a=1;b=2;c=3
result=(all((a<5,b<5,c<5)))
print(result)
```

```python
age=25
nationality="indian"
voting_card=True

if all((age>18,nationality=="indian",voting_card==True)):
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")
```

### 2. ANY ( )

```python
a=1;b=2;c=3
result=(any((a<15,b<5,c<5)))
print(result)
```

```python
age=25
nationality="indian"
voting_card=False

if any((age>18,nationality=="indian",voting_card==False)):
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")
```
