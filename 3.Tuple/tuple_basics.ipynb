## Accessing Tuple Element ( )

#### We can access the tuple using print( ) and also from indexing 

### INDEXING

```python
t8=(10,20,30,50,60,60,98,89,78,87)
print(t8)
print(t8[0])
print(t8[1])
print(t8[2])
print(t8[3])
print(t8[4])

print(t8[-1])
print(t8[-2])
print(t8[-3])
print(t8[-4])
print(t8[-5])
print(t8[-6])
```

### SLICING ( )

#### always plus 1 at the end if in tuple value is 20 than +1 it means 21

```python
t9=(10,20,30,50,60,60,98,89,78,87)

print(t9)
print(t9[0:6:2])
print(t9[1:10:1])
print(t9[2:10:3])
print(t9[3:10:4])
print(t9[4:10:5])
print(t9[0:11])
print(t9[5:11])

print(t9[ : ])  # it will take 0 by default and return all element
```

### LOOPING IN TUPLE 

#### 1. WHILE LOOP( )

#### WHILE loop = "Keep going while..."
#### WHILE = CONDITION IS TRUE...........
### while = condition

```python
no1=(10,20,30,50,40,6,506,5,600)
ind=5
while (ind<len(no1)):
    print(no1[ind])
    ind=ind+1

# while block will terminate after valie's are over
```

#### What this does
#### ind starts at 5, so Python starts from index 5:

####  Index:   0   1   2   3   4   5    6    7    8
#### Value:   10  20  30  50  40  6                          506   5   600
            ↑
         start


### 2. FOR LOOP ( )

```python
tax=(10,20,30,40,50,"vishu",90,50.6)
for i in tax:
    print(i)
```

### 3. Ennumerate ( )

#### Here , we got the index number of each eleement 

```python
t=(10,20,30,40,50,60,60,90,80,4,000,1222)
for index , value in enumerate(t):
    print(index,value)
```

```python
t=(10,20,30,40,50,60,60,90,80,True,False,"vishu",1+2j)
for index , value in enumerate(t):
    print(index,value)
```
