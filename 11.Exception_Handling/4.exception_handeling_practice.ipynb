### Exception Handeling :

##### The code first goes into the try block. If an error occurs, Python immediately stops the try block and moves to the matching except block.

#### you get a error when you try to divde number by zero and it's a zero division errror 

```python
# try → try the code

# error happens → leave try

# except → handle the error
```

#### 1. ZeroDivisionError

```python
try:

    a=10
    b=0
    print(a/b)

except ZeroDivisionError:
    print("Error  Division by zero is not allowed ")
```

#### 2. OverFlowError : 

#### overflow errror happens when number is too large to calculate a math operation

```python
import math 
try:
    print(math.exp(1000))
except OverflowError:
    print("number is too large to calculate the value")
```

#### 3. TypeError 

##### Type errror occurec when you try to do a operation in wrong data type

```python

try:
   a=100
   b="vaishnavi"
   print(a+b)
   
except TypeError:
   print("this is an error because we dont cacluate string and integer value")
```

#### 5. Module NotFound error 

#### This gaves an error when python try to import a module that it cannot find 

```python
try:
    import sparct
except ImportError:
    print("err")
```

#### 6. SyntaxError

```python
try:
    print("Hello):
except SyntaxError:
    print("Error")
```

#### 7. Value errror

#### ValueError happens when Python gets the right type of data, but the value is not acceptable

```python
try:
    a=10
    b="hello world"
    print(int(b))
except ValueError:
    print("value err")
```

### 8. IndexError

#### IndexError happens when you try to access a position (index) that doesn't exist in a list.

```python

try:
    a=(10,20,30,40,50)
    print(a[6])
except IndexError as err:
    print("error")    
```

#### 9.AttributeError

#### AttributeError happens when you try to use an attribute or method that an object doesn't have.

```python
try:
    a="vaishnavi"
    a.append("aher")
    print(a)
except AttributeError as err:
    print("error")
```

#### 10.Key Error

#### KeyError happens when you try to get a key from a dictionary that doesn't exist.

```python
try:
    a={"a":10 , "b":50, "c":30}
    print(a["f"])
except KeyError as err:
    print("error")
```
