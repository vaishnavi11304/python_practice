### 5. ANY ( )

#### Any Function Returns True if anyone element is true in list


```python
ad=[0,0,0,0,100]
print(any(ad))
```

#### We consider 0 as a False , and 1 is Consider As a True

```python
ad=[0,0,0,0,]
print(any(ad))
```

### 6. ALL ( )

#### all function returns true , if all element are true

```python
a5=[1,2,3,5,4]
print(all(a5))
```

```python
a6=[0,1,2,5,6]
print(all(a6))
```
