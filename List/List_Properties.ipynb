### 6.Clonning [ ]  - Deep Copy

#### in clonning We Copy Contents Of One List Into Another List after that both are diffrent list and there memory address are also diffrenet , if we change 1 list it not change anotherb list it,s also called deep copy ()

```python
l1=[1,2,3,4,5,6,7,8]
l2=l1.copy()
l3=l2.copy()
print(l2)
print(l3)
print(id(l2))
print(id(l3))

l4=l3.copy()
print(l4)

l4.remove(8)
print(l4)
```
