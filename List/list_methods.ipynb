## LIST METHOD [ ]

### 1. Adding element is the list ( )

#### i. APPEND ( )

#### This method add the element to the end of the lsit

```python
a=[10,20,30,52,266]
a.append(633)
print(a)
```

```python
a2=[10,20,30,52,266]
a2.append(26)
print(a)
```

#### ii . Insert ( )

#### Using this method we can add or insert particular element index or poisition

```python
w=[10,20,30,4,5,5,9,6]
print(w)
w.insert(0,26)
print(w)
```

```python
wx=[10,20,30,4,5,5,9,6]
print(wx)
wx.insert(0,"vaishnavi")
print(wx)
```

### iii. Extend ( )

#### Using this method we add the 2nd list to the end of another list like concatenation

```python
a12=[10,23,63]
a13=[20,56,56,89]
a12.extend(a13)
print(a12)
```

```python
a14=[10,23,63]
a15=[20,56,56,89,56,1.6,89632655]
a14.extend(a15)
print(a14)
```

### 2. Removing Element from The list 

#### i. remove ( )

#### it removes the particular element from the list

```python
a=[10,20,30,56,90]
a.remove(30)
print(a)
```

```python
b=[10,20,30,56,90,"vaishu"]
b.remove("vaishu")
print(b)
```

### ii. Pop ( )

#### in this pop ( ) method we remove element from the end of the list

```python
a=[10,20,30,56,90]
a.pop()
print(a)
```

#### annd using this pop(index_number) method we remove element from the particular indexing

```python
a1=[10,20,30,56,90]
a1.pop(2)
print(a1)
```
