### BUILT-IN-FUNCTION - tuple ( )

#### 1.LEN ( )

##### It Returns Count of Element In The Tuple

```python
a=(10,20,30)
print(len(a))
```

```python
b=(60,90,80,1,2,3.6,1,2)
print(len(b))
print(a[0])
```

#### 2. MAX ( )

##### It Return largest value available in tuple

```python
b=(60,90,80,1,2,3.6,1,2)
print(max(b))
print(a[0])
```

```python
a=(10,20,30)
print(max(a))
```

### 3. MIN ( )

#### it return Smallest value availabel in tuple

```python
a=(10,20,30)
print(min(a))
```

```python
b=(60,90,80,1,2,3.6,1,2)
print(min(b))
print(a[0])
```

### 5.ANY ( )

##### it's return true if any one element is True in the tuple

```python
f=(0,0,1)
print(any(f))
```

```python
f=(0,0,0)
print(any(f))
```

### 6. ALL ( ) 

##### It returns true if all the elements of the tuple is false , if any one element is flase it return false

```python
a=(10,20.3,"ganesh",True,90)
print(all(a))
```

```python
b=(0,0,0)
print(all(b))
```

```python
c=(10,1)
print(all(c))
```

### 7. SORTED ( )

##### It Return Tuple in Sorted ( Ascending Order)

```python
t1=(60,90,10,1,5,6,0,80)
print(sorted(t1))
```

```python
f=(0,0,1)
print(sorted(f))
```
