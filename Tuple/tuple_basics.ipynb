#### Tuple- Tuple is Container Data Types 

#### Using round braces - seprated by commas

```python
tu=(1,2,3,5)
print(tu)
print(type(tu))
```

### Homogeneous - only 1 data types

```python
a=(1,2,3,5)
b=("vishu","aher")
print(a)
print(b)
print(type(a))
```

### Hetrogeneous - multiple data types

```python
a=(1,2,3,45.6,4+5j,"vaishu",True,False)
print(a)
```

### Creating empty tuple

```python
c=()
print(c)
print(type(c))
```

#### If we Have only 1 value inside tuple

#### if we add comma at the end of the element it consider it is tuple , otherwise it consider int

```python
d=(5,)
print(d)
print(type(d))
```

```python
a69=(5)
print(a69)
print(type(a69))
```

### Duplicates are allowed

```python
b=(10,20,1,5,1,10,60,2000)
print(b)
```

## Accessing Tuple Element ( )

#### We can access the tuple using print( ) and also from indexing 

```python
t8=(10,20,30,50,60,60,98,89,78,87)
print(t8)
print(t8[0])
print(t8[1])
print(t8[2])
print(t8[3])
print(t8[4])

print(t8[-1])
print(t8[-2])
print(t8[-3])
print(t8[-4])
print(t8[-5])
print(t8[-6])
```
