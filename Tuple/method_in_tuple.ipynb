### TUPLE - METHOD

### 1. INDEX ( )

#### It returns the index no.of particular elements of the tuple

##### it gives value error if we gave those element which is not avaible in tuple

```python
t1=(10,20,30,40,65,"vishu")
print(t1.index(30))
print(t1.index(10))
print(t1.index(20))
print(t1.index(40))
print(t1.index(65))
print(t1.index("vishu"))
```

```python
t2=(10,20)
print(t2.index(30))
```

### 2. COUNT ( )

#### It return the aaperance of particuar element in the tuple

#### it check how many time particular element is present in tuple

```python
t1=(10,20,30,40,65,10,20,40,"vishu")
print(t1.count(10))
print(t1.count(20))
print(t1.count(30))
print(t1.count(40))
print(t1.count(65))
print(t1.count("vishu"))
```

##### if particluar element is not present in the tuple it's return 0

```python
t2=(10,20)
print(t2.count(30))
```

### NESTED TUPLE - tuples in tuple

#### using this method we can create so many tuple inside the tuples

```python
a=(10,20,30)
b=(20,90,100)
c=(a,b)
print(c)
print(c[0])
print(c[0][0])
print(c[0][2])
print(c[0][1])
print(c[1][0])
print(c[1][1])
print(c[1][2])
```
