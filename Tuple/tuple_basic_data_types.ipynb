### Basic Data Types Of Tuple ( )

### 1. Immutable 

##### Once its create it cannot be change , modify , or remove the element  

```python
t1=(10,20,30,40)
t1[0]=11
print(t1)
```

#### 2&3 . CONACTENATION ( ) & MERGING - Both are same

#### we can join one tuple at the end of another tuple using '+' operator

```python
a=(10,20,30,40,50,60)
b=(50,60,80,89)
c=a+b
print(c)
```

```python
d1=(12,123)
e1=(13,134)
f1=d1+e1
print(f1)
```

### 4 . CONVERSION - ( )

#### i . converting string to tuple ()

```python
s="hello"
t=tuple(s)
print(t)
print(type(t))
```

#### ii . converting list to tuple

```python
l1=[10,20,30]
t2=tuple(l1)
print(t2)
print(type(t2))
```

#### iii . converting set into tuple - ( )

```python
s1={10,20.6,"vishu"}
s2=tuple(s1)
print(s2)
print(type(s2))
```

#### iv . converting dictionary to tuple - ( )

#### if we convert dictionary with any data type here (we are convrting dict to tuple) it will return only key from dictionary

```python
d1={"a":1 , "b":2}
t1=tuple(d1)
print(t1)
print(type(t1))
```

#### v . converting range to tuple - ( ) 

#### Range function we seperate by comma's not from this -  ( : )

```python
r1=range(1,11,1)
t2=tuple(r1)
print(t2)
print(type(t2))
```
