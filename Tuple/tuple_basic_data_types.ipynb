### Basic Data Types Of Tuple ( )

### 1. Immutable 

##### Once its create it cannot be change , modify , or remove the element  

```python
t1=(10,20,30,40)
t1[0]=11
print(t1)
```

### 4 . CONVERSION - ( )

#### i . converting string to tuple ()

```python
s="hello"
t=tuple(s)
print(t)
print(type(t))
```

#### ii . converting list to tuple

```python
l1=[10,20,30]
t2=tuple(l1)
print(t2)
print(type(t2))
```

#### iii . converting set into tuple - ( )

```python
s1={10,20.6,"vishu"}
s2=tuple(s1)
print(s2)
print(type(s2))
```

#### iv . converting dictionary to tuple - ( )

#### if we convert dictionary with any data type here (we are convrting dict to tuple) it will return only key from dictionary

```python
d1={"a":1 , "b":2}
t1=tuple(d1)
print(t1)
print(type(t1))
```

#### v . converting range to tuple - ( ) 

#### Range function we seperate by comma's not from this -  ( : )

```python
r1=range(1,11,1)
t2=tuple(r1)
print(t2)
print(type(t2))
```

### 7. SEARCHING -'in' operator ( )

#### Using In Operator We can Check Weather Particular Element Is Present In tuple or not if present it's return True if not - False

```python
a=(10,20,30,"vishu")
print(10 in a)
print(20 in a)
print(30 in a)
print("vishu" in a)
print(60 in a)
```

```python
b=(True , 3+5j , False, 56)
print(True in b)
print(False in b)
print(3+5j in b)
print(56 in b)
print(569 in b)
print(9895521 in b)
```

### 8 . Identity - 'is' Operator

#### using this operator it's Check Weather two variables are refer to the same memory location or not if refer it's True , Else - false 

```python
a=(10,20,30)
b=a
print(a is b)

```

### 9. COMPARISON

##### we can compare content of tuples , we compare it by element wise

```python
a1=(10,20,30,40)
b1=(10,20,30,40,50)
print(a1>b1)
print(a1<b1)
```
