## LIST METHOD [ ]

### 1. Adding element is the list ( )

#### i. APPEND ( )

#### This method add the element to the end of the lsit

```python
a=[10,20,30,52,266]
a.append(633)
print(a)
```

```python
a2=[10,20,30,52,266]
a2.append(26)
print(a)
```

#### ii . Insert ( )

#### Using this method we can add or insert particular element index or poisition

```python
w=[10,20,30,4,5,5,9,6]
print(w)
w.insert(0,26)
print(w)
```

```python
wx=[10,20,30,4,5,5,9,6]
print(wx)
wx.insert(0,"vaishnavi")
print(wx)
```

### iii. Extend ( )

#### Using this method we add the 2nd list to the end of another list like concatenation

```python
a12=[10,23,63]
a13=[20,56,56,89]
a12.extend(a13)
print(a12)
```

```python
a14=[10,23,63]
a15=[20,56,56,89,56,1.6,89632655]
a14.extend(a15)
print(a14)
```

### 2. Removing Element from The list 

#### i. remove ( )

#### it removes the particular element from the list

```python
a=[10,20,30,56,90]
a.remove(30)
print(a)
```

```python
b=[10,20,30,56,90,"vaishu"]
b.remove("vaishu")
print(b)
```

### ii. Pop ( )

#### in this pop ( ) method we remove element from the end of the list

```python
a=[10,20,30,56,90]
a.pop()
print(a)
```

#### annd using this pop(index_number) method we remove element from the particular indexing

```python
a1=[10,20,30,56,90]
a1.pop(2)
print(a1)
a1.pop(0)
print(a1)
```

### 7.INDEX( )

#### Using index method we can print index of that particular element

```python
a1=[10,20,30,56,90,40,50,60,70]
print(a1.index(10))
print(a1.index(20))
print(a1.index(30))
print(a1.index(40))
print(a1.index(50))
print(a1.index(60))
print(a1.index(70))
print(a1.index(56))
```

### 8. COUNT ( )

#### We use this method to print how many time particular element appears in the list

```python
a1=[10,20,30,56,90,40,50,60,70,10,20,30]
print(a1.count(10))
print(a1.count(20))
print(a1.count(30))
print(a1.count(40))
print(a1.count(50))
print(a1.count(60))
print(a1.count(70))
print(a1.count(100))
```

### 9. SORT ( )

#### using this method we can rearrenage the element by ascending order 

```python
a1=[10,20,30,56,90,40,50,60,70,10,20,30]
a1.sort()
print(a1)
```

```python
a2=[1,5,9,100,65,86]
a2.sort()
print(a2)
```

```python
a3=[55,65,11,25,5,101]
a3.sort()
print(a3)
```

### 10 . REVERSE ( )

#### using this method , we can print the element of the list in reverse order.

```python
a3=[55,65,11,25,5,101]
a3.reverse()
print(a3)
```

```python
a2=[1,5,9,100,65,86]
a2.reverse()
print(a2)
```

```python
a1=[10,20,30,56,90,40,50,60,70,10,20,30]
a1.reverse()
print(a1)
```

### 11. COPY ( )

#### using this method we can create a another list and create it but the memory address are diffrent

```python
a1=[10,20,30,56,90,40,50,60,70,10,20,30]
a6=a1.copy()
print(a6)
print(id(a6))
print(id(a1))
print(a1)
```
