### Built-in-functions ( ) - on set's

#### 1. LEN( ) - it's return the count of length 

```python
a={10,2,2,5,4,8,96,69}
print(len(a))
```

```python
a1={1,52,63,1}
print(len(a1))
```

#### 2. MA ( ) - it's return the largest value from the set 

```python
b={10,20,30,50,800}
print(max(b))
```

```python
b1={1,2,3}
print(max(b1))
```

#### 3. MIN ( ) - return the smallest value

```python
c={10,20,30,50,800}
print(min(c))
```

```python
c1={1,2,3}
print(min(c1))
```

#### 4. SUM ( ) - Addition from all available set 

```python
d={10,20,30,50,800}
print(sum(d))
```

```python
d1={1,2,3}
print(sum(d1))
```

#### 5.ANY ( ) - it's check any afrom above set is true or, not if any one is true it's return TRUE , ele FALSE

```python
f={10,20,30,50,800, 0, 0}
print(any(f))
```

```python
f1={0,0,2,3}
print(any(f1))
```

#### 6. ALL ( ) - it's check all are true oor not


```python
g={10,20,30,50,800, 0, 0}
print(all(g))
```

```python
g1={1,2,3}
print(all(g1))
```

#### 7. SORTED ( ) - it's sort the et in ascending Order 

```python
h={10,20,30,50,800, 0, 0}
print(sorted(h))
```

```python
h1={1,52,63,1}
print(sorted(h1))
```
