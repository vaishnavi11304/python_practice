### 2. IF-ELSE

```python
age=25
if (age>=18):
    print("you are eligible for voting")
else:
    print("you are not eligible for voting")

print("statement executed")
```

### If-Elif-Else ()

```python
age=20

if (age<13):
    print("you are a child")
elif (age<20):
    print("you are teenager,not eligible for voting")
else:
    print("you are eligible for voting")
    
```
