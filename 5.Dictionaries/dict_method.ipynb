#### Dictionary Method

### 1 . Update ( ) 

##### We can add Multiple Key-value pair or update exiting Key-Value Pairs

##### We can Update Using Two Method

```python
d1={10:"v",
    20:"a"}
d1.update({60:"i"})
print(d1)
```

```python
d1={10:"v",
    20:"a"}
d1["k"]=5
print(d1)
```

#### 2. POP ( )

##### We Provide key to the POP method & it delete particular pair

```python
d1={10:"v",
    20:"a"}
d1.pop(20)
print(d1)
```

```python
d1={10:"v",
    20:"a",
    30:"k",
    60:"n"}
d1.pop(60)
d1.pop(10)
d1.pop(30)
print(d1)
```

#### 3. POP ITEM ( )

##### it's delete the last key-value pair from the dictionary

```python
d1={10:"v",
    20:"a",
    30:"k",
    60:"n"}
d1.popitem()
print(d1)
```
