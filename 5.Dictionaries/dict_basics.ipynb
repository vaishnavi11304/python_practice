#### * Accessing Dictionary Elements 

#### Indexing and slicing does not work in dict

```python
details={"name":"vaishnavi",
         "working":"bajel_pvt ltd",
         "emp code":232757}
print(details)
print(details["working"])
print(details["name"])
print(details["emp code"])

# details.keys()    # only keys
# details.values()  # only values
# details.items()   # key + value
# details["name"]   # value of a specific key

print(details.keys())
print(details.values())
print(details.items())
print(details["name"])
```

```python
a={"a":10,
   "b":10,
   "c":20,"d":40,
   "a":50}
print(a)
print(a["a"])
print(a["b"])
print(a["c"])
print(a["d"])
```

### Looping In Dictionary

#### 1. FOR LOOP

```python
details={"name":"vaishnavi",
         "working":"bajel_pvt ltd",
         "emp code":232757}
for k,v in details.items():
    print(k,v)
```

##### If we want space or any other symbol we use (" ") this in between k and v   

#### Here we accessing key-value both

```python
details={"name":"vaishnavi",
         "working":"bajel_pvt ltd",
         "emp code":232757}
for k,v in details.items():
    print(k,"-" ,v)
```

#### Here , we accessing keys only

```python
details={"name":"vaishnavi",
         "working":"bajel_pvt ltd",
         "emp code":232757}
for k in details.keys():
    print(k)
```

#### Here we accessing Values

```python
details={"name":"vaishnavi",
         "working":"bajel_pvt ltd",
         "emp code":232757}
for v in details.values():
    print(v)
```

#### In Dictionary We Use k for - keys , and V for - values

### ENUMERATE ( ) 

#### We Print Index of dict using enumerate

```python
details={"name":"vaishnavi",
         "working":"bajel_pvt ltd",
         "emp code":232757}
for i, (k,v) in enumerate(details.items()):
    print(i,k,"- ",v)
```
