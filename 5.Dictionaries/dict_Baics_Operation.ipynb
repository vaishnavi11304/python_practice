### Dictionary Basic's Operation

### 1 . Mutability ( )

#### We can , chnage, modify,update,delete add it 

```python
details={"name":"vaishnavi",
         "working":"bajel_pvt ltd",
         "emp code":232757}
details["age"]=25
print(details)
```

### 2. Conversion ( )

#### i . Converting Tuple To Dictionary

```python
add=(("name","vishu"),("pin",423601))
print(type(add))
add[1][0]
add2=dict(add)
print(add2)
print(type(add2))
```

#### ii .  converting list to dictionary ( ) - list of tuples

```python
name_list=[("vaishnavi","vishu"),("mau","samadhan")]
change=dict(name_list)
print(change)
print(type(change))
```

```python
name_list1=[("vaishnavi",653),("mau","samadhan"),("friends","friend")]
change2=dict(name_list1)
print(change2)
print(type(change2))
```

#### iii . converting Set To ditionary ( )

```python
new_set=["vishu","sam","vaishnavi"]
dict1={key:102 for key in new_set}
print(dict1)
print(type(dict1))
```

#### iv . Converting range as kkeys & value as a default to dictionary

```python
r1=range(0,10,6)
d1={key:10 for key in r1}
print(d1)
print(type(d1))
```

#### 3. Aliasing ( )

#### Multiple variable refer to the same memory address

```python
d2={"vishu":10 , "sam":200}
d3=d2
print(type(d3))
print(d3)
print(id(d3))
print(id(d2))
```

#### 4. Clonning ( )

```python
d2={"vishu":10 , "sam":200}
d3=d2.copy()
print(type(d3))
print(d3)
print(id(d2))
print(id(d3))
```

#### 5. Searching "in" Operator

```python
d2={"vishu":10 , "sam":200}
print("vishu" in d2)
print("samadhan" in d2)
```

#### 6. Identity - "IS" Operator

```python
d2={"vishu":10 , "sam":200}
d3=d2
print(type(d3))
print(d3)
print(id(d3))
print(id(d2))
print(d3 is d2)
print(d2 is d3)
d4=d3.copy
print(d4 is d3)
```
