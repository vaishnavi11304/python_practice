### BUILT-IN-FUNCTION

#### 1. LEN ( )

```python
d2={"vishu":10 , "sam":200}
print(len(d2))
```

#### 2. MIN ( )

####  For min and max we always consider the keys

```python
d2={"a":10 , "b":20 , "c": 40}
print(min(d2))
```

#### 3.MAX( )

```python
d2={"a":10 , "b":20 , "c": 40}
print(max(d2))
```

#### 4.SUM ( )

```python
d2={10:"a",
    20:"b",
    30:"c",
    60:"d"}
print(sum(d2))
```

#### 5.ANY( )

#### any madhe ek tri key true pahije

```python
d2={"vishu":10 , "sam":200,0:"h"}
print(any(d2))
```

#### 6.ALL ( ) 

#### all keys need to True if all is not True it return false

```python
d2={"vishu":10 , "sam":200,0:"b"}
print(all(d2))
```

#### 7. SORTED ( )

```python
d2={60:"b",
    10:"m"}
for key in sorted(d2):
    print(key)
```

#### Using this method we can sorted it

```python
d2={60:"b",
    10:"m"}
for key in sorted(d2):
    print(key,":",d2[key])
```

#### 8. REVERSED ()

```python
d2={60:"b",
    10:"m"}
for key in reversed(d2):
    print(key)
```

```python
d2={60:"b",
    900:"u",
    56:"l",
    10:"m"}
for key in reversed(d2):
    print(key,":",d2[key])
```

```python
d2={60:"b",
    900:"u",
    56:"l",
    10:"m"}
print(reversed(d2)) # this gives an object
print(list(reversed(d2)))

# if we use this method we got only keys not values too
```
