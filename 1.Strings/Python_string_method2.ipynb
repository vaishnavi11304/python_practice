## String Method

### 1.Replace

```python
str1="my name is jay"
print(str1)
new_str1=str1.replace("jay","vaishnavi")
print(new_str1)
```

## Trims Whitespace

### 1. Lstrip ()

```python
str2="  new assaignmet is here"
print(str2)
new_str2=str2.lstrip()
print(new_str2)
```

### 2.Rstrip ()

```python
str3="new assaignmet is here     "
print(str3)
new_str4=str3.rstrip()
print(new_str4) 
```

### 3. Strip ()

```python
str4="   new assaignmet is here     "
print(str4)
new_str5=str4.strip()
print(new_str5) 
```

## Split ()

```python
s1="Hello welcome to python class"
print(s1)
s2=s1.split(" ")
print(s2)
```

```python
str4="hello , welcome to python class"
print(str4)
words=str4.split(" , ")
print(words)
```

## Partition ()

```python
s1="Hello welcome to python class"
print(s1)
before,sep,after=s1.partition(" to")
print(before)
print(sep)
print(after)
```

## String Conversion Method

### 1.Lower ()

```python
a="JAI SHREE KRISHNA"
print(a)
b=a.lower()
print(b)
```

### 2. Upper ()

```python
a="Jai Shree Krishna"
print(a)
b=a.upper()
print(b)
```
